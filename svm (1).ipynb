```python
import numpy as np

class logisticreg:
    def __init__(self, learn_rate, iteration):
        self.learn_rate = learn_rate
        self.iteration = iteration

    def sigmoid(self, x):
        return 1 / (1 + np.exp(-x))

    def fit(self, x, y):
        self.m, self.n = x.shape
        self.weight = np.zeros(self.n)
        self.bias = 0

        for i in range(self.iteration):
            z = np.dot(x, self.weight) + self.bias
            y_pred = self.sigmoid(z)

            dw = (1 / self.m) * np.dot(x.T, y_pred - y)
            db = (1 / self.m) * np.sum(y_pred - y)

            self.weight = self.weight - self.learn_rate * dw
            self.bias = self.bias - self.learn_rate * db

    def predict(self, x):
        z = np.dot(x, self.weight) + self.bias
        v1 = self.sigmoid(z)
        return np.where(v1 >= 0.5, 1, 0)
```

```python
import numpy as np

class svm1:
    def __init__(self, learn_rate, epochs, lambda1):
        self.learn_rate = learn_rate
        self.epochs = epochs
        self.lambda1 = lambda1

    def fit(self, x, y):
        self.m, self.n = x.shape
        y = np.where(y <= 0, -1, 1)

        self.w = np.zeros(self.n)
        self.b = 0

        for epoch in range(self.epochs):
            for i, j in enumerate(x):
                condition = y[i] * (np.dot(j, self.w) + self.b)

                if condition >= 1:
                    dw = 2 * self.lambda1 * self.w
                    db = 0
                else:
                    dw = 2 * self.lambda1 * self.w - (y[i] * x[i])
                    db = -y[i]

                self.w = self.w - self.learn_rate * dw
                self.b = self.b - self.learn_rate * db

    def predict(self, x):
        v = np.dot(x, self.w) + self.b
        return np.sign(v)
```
